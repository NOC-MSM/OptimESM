# UKESM1-2-LL

### **UKESM1-2-LL:** Sea Ice Area

```python
# -- Define filepaths -- #
ukesm1_dir = "/g100_work/optim_IAC/research/noc/otooth/OptimESM/WP5/data/MOHC/siarea"
ukesm1_fnames = "UKESM1-2-LL_*_r1i1p1f1_siarea_*.nc"

ukesm1_fpaths = sorted(glob.glob(f"{ukesm1_dir}/{ukesm1_fnames}"))
ukesm1_fpaths
```

```python
# -- Variable -- #
var_name = 'siarea'

# -- Pre-Processing esm-hist -- #
ds_hist = xr.open_dataset(f"{ukesm1_dir}/UKESM1-2-LL_esm-hist_r1i1p1f1_{var_name}_1850-01-2014-12.nc")
ds_hist['time_counter'] = np.arange(np.datetime64('1850-01'), np.datetime64('2015-01'))
ds_hist = ds_hist.sel(time_counter=slice('1978-01', '2024-12'))

# -- March NH Sea Ice Area -- #
print(ds_hist['siarea_NH'].isel(time_counter=slice(2, None, 12)).mean(dim='time_counter').values / 1E12)
print(ds_hist['siarea_NH'].isel(time_counter=slice(2, None, 12)).std(dim='time_counter').values / 1E12)

# -- September SH Sea Ice Area -- #
print(ds_hist['siarea_SH'].isel(time_counter=slice(8, None, 12)).mean(dim='time_counter').values / 1E12)
print(ds_hist['siarea_SH'].isel(time_counter=slice(8, None, 12)).std(dim='time_counter').values / 1E12)
```
